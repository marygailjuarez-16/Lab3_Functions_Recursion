```python
import functools

# Inputs
LAST_NAME = "JUAREZ"
SEED_NUM = 43
FAVORITE_ARTIST = "Arctic Monkeys"

# Decorator to record execution log
def diagnostic_logger(func):
    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        print(f"[LOG] Executing diagnostic pipeline for engineer {LAST_NAME}...")
        return func(*args, **kwargs)
    return wrapper

# Reusable Functions
def validate_reading(reading):
    try:
        val = float(reading)
        if 0 <= val <= 100:
            return True, val, "Valid"
        return False, val, "Out of operational bounds (0-100)"
    except (ValueError, TypeError):
        return False, reading, "Non-numeric data"

def calculate_status(val):
    if val < 30:
        return "Low Output / Fault Risk"
    elif val <= 85:
        return "Optimal Condition"
    else:
        return "Critical Overload Warning"

@diagnostic_logger
def run_equipment_diagnostics():
    # 1. Generate Student-Specific Readings
    raw_data = [SEED_NUM, len(LAST_NAME) * 10, len(FAVORITE_ARTIST) * 5, "INVALID_ERR", -15, 92]
    
    valid_readings = []
    execution_logs = []
    diagnostic_results = []
    
    # 2. Process and Validate
    for item in raw_data:
        is_valid, val, msg = validate_reading(item)
        if is_valid:
            valid_readings.append(val)
            condition = calculate_status(val)
            diagnostic_results.append(f"Reading {val}: {condition}")
            execution_logs.append(f"SUCCESS: Processed value {val} -> {condition}")
        else:
            execution_logs.append(f"ERROR: Skipped input '{item}' ({msg})")
            
    # 3. Summary
    print("--- EQUIPMENT DIAGNOSTIC REPORT ---")
    print(f"Generated Equipment Data: {raw_data}")
    print(f"Validation Results: {len(valid_readings)} Valid | {len(raw_data) - len(valid_readings)} Invalid")
    print("Diagnostic Results:")
    for res in diagnostic_results:
        print(f"  - {res}")
    print("Execution Log:")
    for log in execution_logs:
        print(f"  - {log}")
    print(f"Final Output: SYSTEM HEALTH SCORE = {sum(valid_readings)/len(valid_readings):.2f}%")

run_equipment_diagnostics()
```

```python
# Inputs
LAST_NAME = "JUAREZ"
SEED_NUM = 43
FAVORITE_ARTIST = "Arctic Monkeys"

# 1. Generate Fault Code
initial_fault_code = SEED_NUM * len(LAST_NAME) + len(FAVORITE_ARTIST) # 43 * 6 + 14 = 272

# 2. Recursive Fault Trace Function
def trace_fault(fault_code, call_count=1, trace_log=None):
    if trace_log is None:
        trace_log = []
        
    trace_log.append(f"Level {call_count}: Analyzing Fault Code {fault_code}")
    
    # Base Condition: Fault code reduced below safety threshold (< 20)
    if fault_code < 20:
        trace_log.append(f"Base condition reached at level {call_count}. Fault Isolated.")
        return call_count, trace_log, "FAULT_ISOLATED_OK"
    
    # Recursive Call: Step-down fault reduction
    new_code = fault_code // 2
    return trace_fault(new_code, call_count + 1, trace_log)

# Run Trace
total_calls, trace_history, result = trace_fault(initial_fault_code)

# 3. Display Results
print("--- RECURSIVE FAULT TRACE REPORT ---")
print(f"Generated Fault Data: Initial Code = {initial_fault_code}")
print("Recursive Trace History:")
for step in trace_history:
    print(f"  - {step}")
print(f"Number of Recursive Calls: {total_calls}")
print(f"Execution Log: Handled base condition successfully after {total_calls} iterations.")
print(f"Final Output: Diagnostic Status = {result}")
```

```python
import functools

LAST_NAME = "JUAREZ"
SEED_NUM = 43
FAVORITE_ARTIST = "Arctic Monkeys"

# Decorator for Monitoring
def monitor_pipeline(func):
    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        print("[PIPELINE LOG] Starting Telemetry Data Processing...")
        return func(*args, **kwargs)
    return wrapper

# Generator for Telemetry Stream
def telemetry_generator():
    base = [SEED_NUM, 12, len(LAST_NAME)*5, "BAD_SIG", 98, len(FAVORITE_ARTIST)*6, -5]
    for val in base:
        yield val

# Recursive Analysis for Abnormalities
def recursive_abnormal_analysis(value, depth=1):
    if depth >= 3:
        return f"Critical Fault Verified at Depth {depth}"
    return recursive_abnormal_analysis(value, depth + 1)

@monitor_pipeline
def run_pipeline():
    raw_telemetry = []
    valid_readings = []
    invalid_readings = []
    abnormal_conditions = []
    execution_logs = []

    # Lambda for Normalization
    normalize = lambda x: round(x * 1.05, 2)

    # Process Telemetry Stream
    stream = telemetry_generator()
    for item in stream:
        raw_telemetry.append(item)
        try:
            val = float(item)
            if val < 0 or val > 100:
                invalid_readings.append(val)
                execution_logs.append(f"OUT_OF_BOUNDS: Value {val} rejected.")
            else:
                norm_val = normalize(val)
                valid_readings.append(norm_val)
                if norm_val > 80:
                    abnormal = recursive_abnormal_analysis(norm_val)
                    abnormal_conditions.append((norm_val, abnormal))
                    execution_logs.append(f"ABNORMAL DETECTED: {norm_val} -> {abnormal}")
                else:
                    execution_logs.append(f"VALID: Telemetry normalized to {norm_val}")
        except (ValueError, TypeError):
            invalid_readings.append(item)
            execution_logs.append(f"INVALID DATA: '{item}' non-numeric.")

    # Final Output Report
    print("\n--- INTELLIGENT MONITORING PIPELINE REPORT ---")
    print(f"Student Inputs: LAST_NAME={LAST_NAME}, SEED_NUM={SEED_NUM}, ARTIST={FAVORITE_ARTIST}")
    print(f"Generated Telemetry Data: {raw_telemetry}")
    print(f"Valid/Invalid Results: {len(valid_readings)} Valid, {len(invalid_readings)} Invalid")
    print(f"Processed Results: {valid_readings}")
    print(f"Recursive Analysis: {abnormal_conditions}")
    print("Execution Log:")
    for log in execution_logs:
        print(f"  - {log}")
    print(f"\nFinal Output: Total Processed={len(raw_telemetry)} | Valid={len(valid_readings)} | Status=OPERATIONAL")

run_pipeline()
```
